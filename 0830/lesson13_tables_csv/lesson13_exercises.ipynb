## 09｜用 csv.reader 讀取

把兩列 CSV 逐列印出。


```python
import csv
from io import StringIO

csv_text = "name,score\nAmy,90\nJW,100\nCindy,44"
# 請完成
read = csv.reader(StringIO(csv_text))

for row in read:
    print(row)


```

## 10｜用 DictReader 讀取

用欄名取出 Amy 的 score。


```python
import csv
from io import StringIO

csv_text = "name,score\nAmy,90"
# 請完成
read = csv.DictReader(StringIO(csv_text))

for row in read:
    print(row["score"])
```
