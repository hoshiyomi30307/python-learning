## 07｜把營業額由高到低排列

## 08｜計算各城市平均訂單金額

mean 回答的是每筆訂單平均多少錢。

## 10｜size 計算每組資料列數

## 11｜nunique 計算不重複顧客

訂單數與顧客數回答不同問題。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson17_sales.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
customer_count = df.groupby("city")["customer_id"].nunique()
print(customer_count)
```

## 12｜agg 一次做多種統計

同一個營業額欄位可以同時計算總和、平均與最大值。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson17_sales.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
summary = df.groupby("city")["revenue"].agg(["sum", "mean", "max"])
print(summary.round(1))
```

## 13｜使用清楚的統計欄名

命名彙總能讓報表直接表達每一欄的意思。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson17_sales.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
summary = df.groupby("city").agg(
    total_revenue=("revenue", "sum"),
    average_order=("revenue", "mean"),
    order_count=("order_id", "count")
)
print(summary.round(1))
```

## 14｜同時依城市與分類分組

## 15｜reset_index 把索引變回欄位

多層分組後轉回一般表格，比較容易繼續處理。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson17_sales.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
result = df.groupby(["city", "category"])["revenue"].sum()
results = result.reset_index()
print(result.head())
print(results.head())
```

## 16｜篩選彙總後的結果

## 17｜先確認統計問題再選欄位

單價總和通常沒有商業意義，銷售分析更常加總營業額。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson17_sales.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
price_sum = df.groupby("city")["unit_price"].sum()
revenue_sum = df.groupby("city")["revenue"].sum()
print("單價加總：")
print(price_sum)
print("營業額加總：")
print(revenue_sum)
```

## 21｜布林遮罩篩選 NumPy 陣列

條件會先產生 True 與 False，再留下符合的數字。


```python
import numpy as np

prices = np.array([100, 250, 80, 300])
mask = prices >= 200
print(mask)
print(prices[mask])
```

## 22｜NumPy 的加總與平均

陣列可以直接呼叫 sum 與 mean。


```python
import numpy as np

values = np.array([1200, 1800, 950, 2100])
print("總和：", values.sum())
print("平均：", values.mean())
```

## 23｜List 與 ndarray 的乘法不同

List 乘以 2 會重複內容；NumPy 陣列乘以 2 會計算每個元素。


```python
import numpy as np

price_list = [100, 200, 300]
price_array = np.array([100, 200, 300])
print("List：", price_list * 2)
print("NumPy：", price_array * 2)
```

## 24｜Pandas 欄位可以轉成 NumPy 陣列

to_numpy 讓 Series 的數字交給 NumPy 處理。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson17_sales.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
revenue_array = df["revenue"].to_numpy()
print(type(revenue_array))
print("總營業額：", revenue_array.sum())
```

## 25｜完成城市銷售摘要

把新增欄位、分組、統計與排序串成完整流程。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson17_sales.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
report = df.groupby("city").agg(
    total_revenue=("revenue", "sum"),
    average_order=("revenue", "mean"),
    customers=("customer_id", "nunique")
)
report["average_order"] = report["average_order"].round(1)
report = report.sort_values("total_revenue", ascending=False)
print(report)
```
