## Lesson 13｜題目 2：欄位名稱與資料意義

完成 columns，讓每個欄名都對應正確用途，再逐筆印出。


```python
columns = {
    'order_id': '訂單編號',
    'product': '商品名稱',
    'quantity': '購買數量',
    'unit_price': '商品單價',
}
# 逐筆印出欄位名稱與對應用途
#字典走訪：使用 .items() 可以同時取得字典的鍵（key，即欄位名稱）與值（value，即對應用途）。
for key, value in columns.items():
    print(f"{key}: {value}")
```

## Lesson 13｜題目 3：從原始欄位算出營業額

使用 for 逐筆計算 `quantity * unit_price`，把結果存入每筆訂單的 `revenue`，最後印出訂單。


```python
orders = [
    {'product': '拿鐵', 'quantity': 2, 'unit_price': 120},
    {'product': '貝果', 'quantity': 3, 'unit_price': 65},
]

# 使用 for 迴圈逐筆計算並新增 revenue 欄位
for order in orders:
    order['revenue'] = order['quantity'] * order['unit_price']

# 印出更新後的訂單資料
print(orders)
```

## Lesson 13｜題目 4：使用 DictReader 讀取 CSV

讀取本複習提供的 CSV，印出欄名、資料筆數與第一筆資料。提醒：DictReader 讀出的每一列是 Dict。


```python
import csv
from pathlib import Path

csv_path = Path("lesson13_14_review_orders.csv")

# 使用 utf-8-sig 開啟可避免 Windows 產生的 BOM (EF BB BF) 影響欄位名稱
with open(csv_path, mode="r", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)

    # 1. 印出欄名
    print(f"欄名: {reader.fieldnames}")

    # 轉成 list 方便取得總筆數與第一筆資料
    rows = list(reader)

    # 2. 印出資料筆數
    print(f"資料筆數: {len(rows)}")

    # 3. 印出第一筆資料
    if rows:
        print(f"第一筆資料: {rows[0]}")
```

## Lesson 13｜題目 5：CSV 數字欄位需要轉型

CSV 的 quantity 與 unit_price 讀進來是字串。請轉成 int，逐筆計算並累加總營業額。


```python
from pathlib import Path

csv_path = Path("lesson13_14_review_orders.csv")

import csv

with csv_path.open(encoding="utf-8") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

total_revenue = 0
# 請逐筆轉型並累加
# 使用 for 迴圈走訪每一筆資料並計算總營業額
for row in rows:
    quantity = int(row["quantity"])
    unit_price = int(row["unit_price"])
    total_revenue += quantity * unit_price

print(f"總營業額: {total_revenue}")
```

## Lesson 13｜題目 6：用 Dict 累計商品銷量

逐筆讀取 CSV，用 Dict 累加每種商品的 quantity，最後逐項印出商品與總數量。


```python
from pathlib import Path

csv_path = Path("lesson13_14_review_orders.csv")

import csv

with csv_path.open(encoding="utf-8") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

product_quantity = {}
# 請完成累加
for row in rows:
    product = row["product"]
    quantity = int(row["quantity"])

    # 如果商品已經在字典中，就累加數量；若還不在，就初始化為該數量
    if product in product_quantity:
        product_quantity[product] += quantity
    else:
        product_quantity[product] = quantity

# 逐項印出商品與總數量
for product, total in product_quantity.items():
    print(f"{product}: {total}")
```

## Lesson 13｜題目 7：用 Set 找不重複城市

逐筆把 city 加入 Set，最後印出城市數量與排序後的城市。


```python
from pathlib import Path

csv_path = Path("lesson13_14_review_orders.csv")

import csv

with csv_path.open(encoding="utf-8") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

cities = set()
# 請完成
#逐筆將city加入set
for row in rows:
    cities.add(row["city"])
# 印出城市數量與排序後的城市
print(f"城市數量: {len(cities)}")
print(f"排序後的城市: {sorted(cities)}")
```
