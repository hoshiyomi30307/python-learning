## 04｜loc：用標籤或名稱找資料

請依照說明完成：loc 是 label-based indexing。看到 loc，就先看 index 標籤與欄位名稱。


```python
import pandas as pd

df_people = pd.DataFrame(
    {
        "name": ["JW", "Amy", "Bob"],
        "age": [30, 25, 28],
    },
    index=["a", "b", "c"],
)
print(df_people)
print("標籤 a 的整列：")
print(df_people.loc["a"])
print("標籤 a、name 欄:",df_people.loc["a","name"])



# 1. 使用 loc 取得標籤 a 的整列
# 2. 使用 loc 取得標籤 a、name 欄的值
```

## 05｜iloc：用整數位置找資料

請依照說明完成：iloc 是 integer location。列與欄都從位置 0 開始計算。


```python
import pandas as pd

df_people = pd.DataFrame(
    {
        "name": ["JW", "Amy", "Bob"],
        "age": [30, 25, 28],
    },
    index=["a", "b", "c"],
)
print(df_people)
print("第 0 列")
print(df_people.iloc[0])
print("第 0 列、第 1 欄的值:",df_people.iloc[0,1])
# 1. 使用 iloc 取得第 0 列
# 2. 使用 iloc 取得第 0 列、第 1 欄的值
```
